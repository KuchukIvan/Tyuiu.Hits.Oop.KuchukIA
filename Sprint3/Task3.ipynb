<h3 style="color:DodgerBlue">Описание задачи:</h3>
<p>Создать базовый класс <strong>Review</strong> в C#, который будет представлять отзывы о продуктах или услугах. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.</p>

<h3 style="color:DodgerBlue">Требования к базовому классу Review:</h3>
<ul>
    <li><strong>Атрибуты:</strong> ID отзыва (ReviewId), Текст отзыва (Text), Рейтинг (Rating).</li>
    <li><strong>Методы:</strong>
        <ul>
            <li><code>DisplayReview()</code>: метод для отображения отзыва.</li>
            <li><code>RateProduct()</code>: метод для присвоения рейтинга продукту.</li>
            <li><code>GetReviewDetails()</code>: метод для получения деталей отзыва.</li>
        </ul>
    </li>
</ul>

<h3 style="color:DodgerBlue">Требования к производным классаам:</h3>
<ol>
    <li><strong>ОтзывОбслуживания (ServiceReview):</strong>
        <ul>
            <li>Должен содержать дополнительные атрибуты, такие как Дата посещения (VisitDate).</li>
            <li>Метод <code>DisplayReview()</code> должен быть переопределен для включения даты посещения в отображение отзыва.</li>
        </ul>
    </li>
    <li><strong>ОтзывТовара (ProductReview):</strong>
        <ul>
            <li>Должен содержать дополнительные атрибуты, такие как Идентификатор продукта (ProductId).</li>
            <li>Метод <code>RateProduct()</code> должен быть переопределен для связывания рейтинга с конкретным продуктом.</li>
        </ul>
    </li>
    <li><strong>ОтзывУслуги (ServiceReview) (если требуется третий класс):</strong>
        <ul>
            <li>Должен содержать дополнительные атрибуты, такие как Время начала услуги (StartTime).</li>
            <li>Метод <code>GetReviewDetails()</code> должен быть переопределен для отображения времени начала услуги вместе с другими деталями отзыва.</li>
        </ul>
    </li>
</ol>

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте полиморфизм с перекрытием и прегегрузкой методов, а также generic классы

```
using System;
using System.Collections.Generic;
using System.Linq;

public interface ILikeable
{
    void LikeReview();
}

public interface ICommentable
{
    void AddComment(string comment);
}

public class Review
{
    private int reviewId;
    private string text;
    private int rating;
    private string author;
    private DateTime createdAt = DateTime.Now;

    public int ReviewId
    {
        get { return reviewId; }
        set { reviewId = value; }
    }

    public string Text
    {
        get { return text; }
        set { text = value; }
    }

    public int Rating
    {
        get { return rating; }
        set { rating = Math.Clamp(value, 1, 5); }
    }

    public string Author
    {
        get { return author; }
        set { author = value; }
    }

    public DateTime CreatedAt
    {
        get { return createdAt; }
        set { createdAt = value; }
    }

    public Review()
    {
    }

    public Review(int reviewId, string text, int rating, string author)
    {
        this.reviewId = reviewId;
        this.text = text;
        Rating = rating;
        this.author = author;
    }

    public virtual void DisplayReview()
    {
        Console.WriteLine($"Отзыв: {text} (автор: {author})");
    }

    public virtual void RateProduct(int newRating)
    {
        Rating = newRating;
        Console.WriteLine($"Рейтинг {rating}/5 присвоен продукту");
    }

    public void RateProduct(int newRating, string comment)
    {
        RateProduct(newRating);
        Console.WriteLine($"Комментарий к оценке: {comment}");
    }

    public virtual string GetReviewDetails()
    {
        return $"ID: {reviewId}, Текст: {text}, Рейтинг: {rating}, Автор: {author}, Создан: {createdAt:dd.MM.yyyy}";
    }

    public virtual string GetSummary()
    {
        return $"Отзыв №{reviewId}, автор {author}: [{rating}/5] {text}";
    }

    public void InteractWith(Review review)
    {
        Console.WriteLine($"Отзыв ID: {reviewId} взаимодействует с отзывом ID: {review.ReviewId}");
        Console.WriteLine($"Текст связанного отзыва: {review.Text}");

        int diff = Math.Abs(rating - review.Rating);
        if (diff == 0)
            Console.WriteLine("Рейтинги совпадают — мнения согласуются");
        else
            Console.WriteLine($"Рейтинги отличаются на {diff} балл(ов)");
    }
}

public class ServiceReview : Review
{
    private DateTime visitDate;
    private string staffName;
    private int waitMinutes;

    public DateTime VisitDate
    {
        get { return visitDate; }
        set { visitDate = value; }
    }

    public string StaffName
    {
        get { return staffName; }
        set { staffName = value; }
    }

    public int WaitMinutes
    {
        get { return waitMinutes; }
        set { waitMinutes = value; }
    }

    public ServiceReview()
    {
    }

    public ServiceReview(int reviewId, string text, int rating, string author, DateTime visitDate, string staffName)
        : base(reviewId, text, rating, author)
    {
        this.visitDate = visitDate;
        this.staffName = staffName;
    }

    public override void DisplayReview()
    {
        base.DisplayReview();
        Console.WriteLine($"Дата посещения: {visitDate:dd.MM.yyyy}, Сотрудник: {staffName}");
    }

    public override string GetSummary()
    {
        return base.GetSummary() + $" | Ожидание: {waitMinutes} мин";
    }

    public void ConfirmVisit()
    {
        Console.WriteLine($"Посещение {visitDate:dd.MM.yyyy} подтверждено сотрудником {staffName}");
    }
}

public class ProductReview : Review
{
    private int productId;
    private bool isVerified;

    public int ProductId
    {
        get { return productId; }
        set { productId = value; }
    }

    public bool IsVerified
    {
        get { return isVerified; }
        set { isVerified = value; }
    }

    public ProductReview()
    {
    }

    public ProductReview(int reviewId, string text, int rating, string author, int productId)
        : base(reviewId, text, rating, author)
    {
        this.productId = productId;
    }

    public override void RateProduct(int newRating)
    {
        Rating = newRating;
        Console.WriteLine($"Рейтинг {Rating}/5 присвоен продукту с ID: {productId}");
    }

    public override string GetSummary()
    {
        string verified = isVerified ? "да" : "нет";
        return base.GetSummary() + $" | Покупка подтверждена: {verified}";
    }

    public void VerifyPurchase()
    {
        isVerified = true;
        Console.WriteLine($"Покупка товара с ID: {productId} подтверждена");
    }
}

public class ServiceReviewWithTime : Review
{
    private DateTime startTime;
    private int durationMinutes;

    public DateTime StartTime
    {
        get { return startTime; }
        set { startTime = value; }
    }

    public int DurationMinutes
    {
        get { return durationMinutes; }
        set { durationMinutes = value; }
    }

    public ServiceReviewWithTime()
    {
    }

    public ServiceReviewWithTime(int reviewId, string text, int rating, string author, DateTime startTime)
        : base(reviewId, text, rating, author)
    {
        this.startTime = startTime;
    }

    public override string GetReviewDetails()
    {
        return base.GetReviewDetails() + $", Время начала: {startTime:HH:mm}, Длительность: {durationMinutes} мин";
    }
}

public class PremiumProductReview : ProductReview, ILikeable, ICommentable
{
    private int likesCount;
    private List<string> comments = new List<string>();

    public int LikesCount
    {
        get { return likesCount; }
        set { likesCount = value; }
    }

    public List<string> Comments
    {
        get { return comments; }
        set { comments = value; }
    }

    public PremiumProductReview()
    {
    }

    public PremiumProductReview(int reviewId, string text, int rating, string author, int productId)
        : base(reviewId, text, rating, author, productId)
    {
    }

    public void LikeReview()
    {
        likesCount++;
        Console.WriteLine($"Отзыву ID: {ReviewId} поставили лайк (всего лайков: {likesCount})");
    }

    public void AddComment(string comment)
    {
        comments.Add(comment);
        Console.WriteLine($"Новый комментарий к отзыву ID: {ReviewId} — \"{comment}\" (всего комментариев: {comments.Count})");
    }
}

public class ReviewCollection<T> where T : Review
{
    private List<T> items = new List<T>();

    public List<T> Items
    {
        get { return items; }
    }

    public void Add(T review)
    {
        items.Add(review);
    }

    public double GetAverageRating()
    {
        if (items.Count == 0)
            return 0;

        return items.Average(review => review.Rating);
    }
}

Console.WriteLine("=== 1. Отзыв об обслуживании ===");
ServiceReview service = new ServiceReview(1, "Отличный сервис!", 5, "Анна", DateTime.Now, "Дмитрий");
service.WaitMinutes = 10;
service.DisplayReview();
Console.WriteLine(service.GetReviewDetails());
service.ConfirmVisit();

Console.WriteLine("\n=== 2. Отзыв о товаре ===");
ProductReview product = new ProductReview(2, "Хороший товар", 4, "Игорь", 1001);
product.DisplayReview();
product.RateProduct(4);
product.VerifyPurchase();
Console.WriteLine(product.GetReviewDetails());

Console.WriteLine("\n=== 3. Отзыв об услуге со временем ===");
ServiceReviewWithTime serviceTime = new ServiceReviewWithTime(3, "Быстрое обслуживание", 5, "Мария", DateTime.Now);
serviceTime.DurationMinutes = 45;
serviceTime.DisplayReview();
Console.WriteLine(serviceTime.GetReviewDetails());

Console.WriteLine("\n=== 4. Взаимодействие объектов ===");
serviceTime.InteractWith(service);

Console.WriteLine("\n=== 5. Премиум-отзыв (множественное наследование) ===");
PremiumProductReview premium = new PremiumProductReview(4, "Отличный премиум товар", 5, "Олег", 2002);
premium.DisplayReview();
premium.RateProduct(5);
premium.VerifyPurchase();
Console.WriteLine(premium.GetReviewDetails());
premium.LikeReview();
premium.LikeReview();
premium.AddComment("Очень доволен покупкой!");

Console.WriteLine("\n=== 6. Полиморфизм: переопределение (override) ===");
List<Review> reviews = new List<Review> { service, product, serviceTime, premium };
foreach (Review review in reviews)
{
    Console.WriteLine($"{review.GetType().Name} -> {review.GetSummary()}");
}

Console.WriteLine("\n=== 7. Полиморфизм: перегрузка (overload) ===");
product.RateProduct(3);
product.RateProduct(5, "Поменял мнение — товар супер!");

Console.WriteLine("\n=== 8. Generic-класс ===");
ReviewCollection<ProductReview> productReviews = new ReviewCollection<ProductReview>();
productReviews.Add(product);
productReviews.Add(premium);

foreach (ProductReview review in productReviews.Items)
{
    review.DisplayReview();
}
Console.WriteLine($"Отзывов о товарах: {productReviews.Items.Count}, средний рейтинг: {productReviews.GetAverageRating():F1}");

ReviewCollection<Review> allReviews = new ReviewCollection<Review>();
allReviews.Add(service);
allReviews.Add(product);
allReviews.Add(serviceTime);
allReviews.Add(premium);
Console.WriteLine($"Всего отзывов: {allReviews.Items.Count}, средний рейтинг: {allReviews.GetAverageRating():F1}");
```
